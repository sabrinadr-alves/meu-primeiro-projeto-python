```python
produtos = {
    "Produto": ["A", "B", "C"],
    "Custo": [20, 40, 15],
    "Venda": [35, 55, 30],
    "Quantidade": [100, 80, 150]
}
```

```python
import pandas as pd

df = pd.DataFrame(produtos)
df["Lucro"] = df["Venda"] - df["Custo"]
df["Custo Total"] = df["Custo"] * df["Quantidade"]
df["Lucro Total"] = df["Lucro"] * df["Quantidade"]
df["Margem"] = (
       df["Lucro"] / df ["Venda"] * 100
)
df.index = df.index + 1
display(df)
```
